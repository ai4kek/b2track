### MC Simulations

The **generic**  $B^+ B^-$ and $\bar{B^0} B^0$ events using EvtGen requires _`finalstate='mixed'`_ and _`finalstate='mixed'`_ in _`ge.add_evtgen_generator()`_. Note the branching fractions as follows:

- $\Upsilon (4S) \to B^+ \: B^-$ with $\Gamma_i/\Gamma = 51.4 \pm 0.6 %$
- $\Upsilon (4S) \to \bar{B^0} \: B^0$ with $\Gamma_i/\Gamma = 48.6 \pm 0.6 %$
- $B^0 \to J/\Psi \: K_S^0$ with $\Gamma_i/\Gamma = ???%$ 
- $J/\psi(\to e^+e^-)$ with $\Gamma_i/\Gamma = 5.971%$ 
- $J/\psi(\to \mu^+\mu^-)$ with $\Gamma_i/\Gamma = 5.91%$
- $K_S^0(\to\pi^+\pi^+)$ with $\Gamma_i/\Gamma = 69.20%$

However, generic events are huge and make take several hours or days to simulation. One is interested in more specific **signals** _e.g._ $B^0 \to J/\Psi (\mu^+ \mu^-) \: K_S^0 (\pi^+ \pi^-)$


- $\text{BF}(J/\Psi \to \mu^+ \mu^-) = 1.0$
- $\text{BF}(K^0_S \to \pi^+ \pi^-) = 1.0$

or $B^0 \to \phi (K^+ K^-) K^0_S (\pi^+ \pi^- + \pi^0 \pi^0)$ with braching fractions defined as

- $\text{BF}(\phi \to K^+ K^-) = 1.0$
- $\text{BF}(K^0_S \to \pi^+ \pi^-) = 0.7$
- $\text{BF}(K^0_S \to \pi^0 \pi^0) = 0.3$

### Signal Events

For generation signal MC events for $B^0 \to J/\psi (\mu^+ \mu^-) \: K_S^0 (\pi^+ \pi^-)$, we need following decay file for EvtGen:

```bash
# Decay File:
Decay Upsilon(4S)
1.0 B0 anti-B0  VSS_BMIX dm;
Enddecay

Decay B0
1.0 J/psi K_S0 PHSP;
Enddecay
CDecay anti-B0

Decay J/psi
1.0 mu+ mu- VLL;
Enddecay

Decay K_S0
1.0 pi+ pi- PHSP;
Enddecay

End
```
For generation signal MC events for $B^0 \to \phi (K^+ K^-) K^0_S (\pi^+ \pi^- + \pi^0 \pi^0)$, we need following decay file for EvtGen:

```bash
Decay Upsilon(4S)
1.0 B0 anti-B0  VSS_BMIX dm;
Enddecay

Decay B0
1.0 phi K_S0 PHSP;
Enddecay
CDecay anti-B0

Decay phi
1.0 K+ K- VSS;
Enddecay

Decay K_S0
0.7 pi+ pi- PHSP;
0.3 pi0 pi0 PHSP;
Enddecay

End
```

````python
# Create the steering path
main = b2.Path()
````

````python
# Define number of events and experiment number
main.add_module("EventInfoSetter", evtNumList=[10], expList=[0])
````

- _**Generator**: Generate Signal Events_

````python
# add event generator
ge.add_evtgen_generator(
    path=main, finalstate="signal", signaldecfile=b2.find_file("signal_B0_Jpsi_KS0.dec")
);
````

- _**Simulation**: Simulate the Detector Response and the L1 Trigger_

````python
si.add_simulation(path=main);
````

- _**Reconstruction**: Reconstruct Relevant Objects (Tracks, EM Clusters)_

````python
re.add_reconstruction(path=main)
````

- _**Storage**: Create the mDST output file_

````python
mdst.add_mdst_output(path=main, filename="bbarb.root");
````

- _**Event Loop**_

````python
b2.process(path=main)
````

- _**Statistics**: Print out some statistics about the modules execution_

````python
print(b2.statistics)
````
